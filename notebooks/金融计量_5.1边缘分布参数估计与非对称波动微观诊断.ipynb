```python
import pandas as pd
import warnings
from arch import arch_model

warnings.filterwarnings('ignore')

# 1. 读取数据
df = pd.read_excel('data/LOCAL_INPUT')
r_500 = df['r_500'].dropna()
r_1000 = df['r_1000'].dropna()

# 2. 定义纯参数提取函数
def get_gjr_garch_params(series):
    # 设定 GJR-GARCH(1,1) 模型与 Student-t 分布
    am = arch_model(series, mean='Constant', vol='GARCH', p=1, o=1, q=1, dist='studentst')
    res = am.fit(disp='off', show_warning=False)
    
    # 直接提取参数值为浮点数字典
    params = {
        'mu': round(res.params['mu'], 6),
        'omega': round(res.params['omega'], 6),
        'alpha': round(res.params['alpha[1]'], 6),
        'gamma': round(res.params['gamma[1]'], 6),
        'beta': round(res.params['beta[1]'], 6),
        'nu': round(res.params['nu'], 6)
    }
    return params

# 3. 获取并打印参数
params_500 = get_gjr_garch_params(r_500)
params_1000 = get_gjr_garch_params(r_1000)

print("=== 中证500 模型参数 ===")
for key, value in params_500.items():
    print(f"{key}: {value}")

print("\n=== 中证1000 模型参数 ===")
for key, value in params_1000.items():
    print(f"{key}: {value}")
```
